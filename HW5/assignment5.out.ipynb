### 2.1: K Means ###

```python

################################
# Load the data
################################
import os
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs, make_moons

# Create / load the datasets:
n_samples = 1500
X0, _ = make_blobs(n_samples=n_samples, centers=2, n_features=2, random_state=0)
X1, _ = make_blobs(n_samples=n_samples, centers=5, n_features=2, random_state=0)

random_state = 170
X, y = make_blobs(n_samples=n_samples, random_state=random_state, cluster_std=1.3)
transformation = [[0.6, -0.6], [-0.2, 0.8]]
X2 = np.dot(X, transformation)
X3, _ = make_blobs(n_samples=n_samples, cluster_std=[1.0, 2.5, 0.5], random_state=random_state)
X4, _ = make_moons(n_samples=n_samples, noise=.12)

X = [X0, X1, X2, X3, X4]
# The datasets are X[i], where i ranges from 0 to 4


################################
# Code to plot clusters
################################
def plot_cluster(ax, data, cluster_assignments):
    '''Plot two-dimensional data clusters
    
    Parameters
    ----------
    ax : matplotlib axis
        Axis to plot on
    data : list or numpy array of size [N x 2] 
        Clustered data
    cluster_assignments : list or numpy array [N]
        Cluster assignments for each point in data

    '''
    clusters = np.unique(cluster_assignments)
    n_clusters = len(clusters)
    for ca in clusters:
        kwargs = {}
        if ca == -1:
            # if samples are not assigned to a cluster (have a cluster assignment of -1, color them gray)
            kwargs = {'color':'gray'}
            n_clusters = n_clusters - 1
        ax.scatter(data[cluster_assignments==ca, 0], data[cluster_assignments==ca, 1],s=5,alpha=0.5, **kwargs)
        ax.set_xlabel('feature 1')
        ax.set_ylabel('feature 2')
        ax.set_title(f'No. Clusters = {n_clusters}')
        ax.axis('equal')

# scatter plot each dataset
fig, axes = plt.subplots(1, 5, figsize=(22, 4))
for i, data in enumerate(X):
    axes[i].scatter(data[:, 0], data[:, 1], s=5, alpha=0.5)
    axes[i].set_title(f"Dataset {i}")
    axes[i].set_xlabel("feature 1")
    axes[i].set_ylabel("feature 2")
    axes[i].axis("equal")
plt.tight_layout()
plt.show()

from sklearn.cluster import KMeans, DBSCAN, SpectralClustering
from sklearn.metrics import silhouette_score

k_values = range(1, 11)
inertias = {}

fig, axes = plt.subplots(1, 5, figsize=(22, 4))
for i, data in enumerate(X):
    dataset_inertias = []
    for k in k_values:
        km = KMeans(n_clusters=k, random_state=0, n_init=10)
        km.fit(data)
        dataset_inertias.append(km.inertia_)
    inertias[i] = dataset_inertias
    
    axes[i].plot(list(k_values), dataset_inertias, marker='o')
    axes[i].set_title(f"Dataset {i} Elbow Curve")
    axes[i].set_xlabel("k")
    axes[i].set_ylabel("inertia")
plt.tight_layout()
plt.show()

from sklearn.metrics import silhouette_score
from sklearn.cluster import KMeans

k_values = range(2, 11)  # silhouette not defined for k=1
silhouette_scores = {}

fig, axes = plt.subplots(1, 5, figsize=(22, 4))

for i, data in enumerate(X):
    scores = []
    
    for k in k_values:
        km = KMeans(n_clusters=k, random_state=0, n_init=10)
        labels = km.fit_predict(data)
        score = silhouette_score(data, labels)
        scores.append(score)
    
    silhouette_scores[i] = scores
    
    axes[i].plot(list(k_values), scores, marker='o')
    axes[i].set_title(f"Dataset {i} Silhouette")
    axes[i].set_xlabel("k")
    axes[i].set_ylabel("score")

plt.tight_layout()
plt.show()

best_k_silhouette = {}

for i in range(len(X)):
    scores = silhouette_scores[i]
    best_k = list(k_values)[np.argmax(scores)]
    best_k_silhouette[i] = best_k

print(best_k_silhouette)

fig, axes = plt.subplots(1, 5, figsize=(22, 4))

for i, data in enumerate(X):
    k = best_k_silhouette[i]
    km = KMeans(n_clusters=k, random_state=0, n_init=10)
    labels = km.fit_predict(data)
    
    plot_cluster(axes[i], data, labels)
    axes[i].set_title(f"Dataset {i}, k = {k}")

plt.tight_layout()
plt.show()

```

**Interpretation:** 

From the elbow curves for each dataset (dataset 0, 1, 2, 4), the elbow appears at k=2 (except for dataset 3 which performed better with k=3 clusters), which indicates this is where the inertia begins to have diminishing returns as we increase number of clusters, and this value of k could be a good balance.


### 2.2: DBSCAN ###

```python
# Apply DBSCAN. Vary the eps and min_samples parameters to get as close as you can to having the same number of clusters as your choices with K-means. The same code plots as gray/black any points that were not assigned to clusters.
from sklearn.cluster import DBSCAN

eps_values = np.linspace(0.1, 1.5, 15)
min_samples_values = [3, 5, 10, 15]

best_dbscan_params = {}
best_dbscan_labels = {}

# search for the best dbscan params to get as close to number of clusters as above
for i, data in enumerate(X):
    
    # number of cluster choices for each dataset from kmeans
    target_k = best_k_silhouette[i]
    best_diff = float("inf")
    best_params = None
    best_labels = None
    
    for eps in eps_values:
        for min_samples in min_samples_values:
            db = DBSCAN(eps=eps, min_samples=min_samples)
            labels = db.fit_predict(data)
            
            n_clusters = len(set(labels)) - (1 if -1 in labels else 0)
            diff = abs(n_clusters - target_k)
            
            if diff < best_diff:
                best_diff = diff
                best_params = (eps, min_samples, n_clusters)
                best_labels = labels
    
    best_dbscan_params[i] = best_params
    best_dbscan_labels[i] = best_labels

for i in range(len(X)):
    eps, min_samples, n_clusters = best_dbscan_params[i]
    print(f"Dataset {i}: target k = {best_k_silhouette[i]}, eps = {eps:.2f}, min_samples = {min_samples}, DBSCAN clusters = {n_clusters}")
```

```python
fig, axes = plt.subplots(1, 5, figsize=(22, 4))
fig.suptitle("DBSCAN")

for i, data in enumerate(X):
    plot_cluster(axes[i], data, best_dbscan_labels[i])
    eps, min_samples, n_clusters = best_dbscan_params[i]
    axes[i].set_title(f"Dataset {i}\neps={eps:.2f}, min_samples={min_samples}")

plt.tight_layout()
plt.show()
```

### 2.3: Spectral Clustering ###


```python
#Apply Spectral Clustering. Select the same number of clusters as selected by k-means.
from sklearn.cluster import SpectralClustering

sc_labels = {}


for i, data in enumerate(X):
    
    # number of cluster choices for each dataset from kmeans
    target_k = best_k_silhouette[i]
    
    sc = SpectralClustering(
    n_clusters=target_k,
    affinity='nearest_neighbors',
    n_neighbors=10,
    random_state=42
)
    labels = sc.fit_predict(data)
    sc_labels[i] = labels
    
    
fig, axes = plt.subplots(1, 5, figsize=(22, 4))
fig.suptitle("Spectral Clustering")

for i, data in enumerate(X):
    plot_cluster(axes[i], data, sc_labels[i])
    axes[i].set_title(f"Dataset {i}\n")

plt.tight_layout()
plt.show()
```

**Interpretation:**

And DBSCAN required the most tuning for eps and min_samples, which are sensitive. Spectral Clustering seemed to be consistent across all datasets, as it performed similar to Kmeans on the simple data while also doing well on the non-linear data structures as well. KMeans is the easiest but least flexible, while DBSCAN is the hardest to tune and good for complex shapes, but overall, Spectral Clustering has the best balance.

```python
from sklearn.preprocessing import StandardScaler
# standardize before PCA / t-SNE
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_digits)

# pca to 2d
pca_2d = PCA(n_components=2, random_state=42)
X_pca_2d = pca_2d.fit_transform(X_scaled)

plt.figure(figsize=(9, 7))
scatter = plt.scatter(X_pca_2d[:, 0], X_pca_2d[:, 1], c=y_digits, cmap="tab10", s=18, alpha=0.8)
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("Digits dataset projected to 2D with PCA")
plt.legend(*scatter.legend_elements(), title="Digit")
plt.show()
```

### 3.2 ###

```python
# cumulative explained variance
pca_full = PCA().fit(X_scaled)
cum_var = np.cumsum(pca_full.explained_variance_ratio_)

plt.figure(figsize=(8, 5))
plt.plot(range(1, len(cum_var) + 1), cum_var, marker="o")
plt.xlabel("Number of principal components")
plt.ylabel("Cumulative explained variance ratio")
plt.title("PCA cumulative explained variance")
plt.grid(True, alpha=0.3)
plt.show()

unexplained_2pc = 1 - cum_var[1]
print("Fraction unexplained by first 2 PCs:", unexplained_2pc)
```

The cumulative explained variance plot shows how much total variance is captured as more principal components are added. The fraction of variance unexplained by the first two principal components is 1 - cumulative variance at 2 PCs, which here is the printed value above (0.784). Since a substantial fraction of the variance remains unexplained, the 2D PCA visualization cannot perfectly separate the digits, so overlap between classes is expected.

### 3.3 ###

```python
# t-SNE to 2D
tsne = TSNE(n_components=2, perplexity=30, random_state=42, init="pca", learning_rate="auto")
X_tsne_2d = tsne.fit_transform(X_scaled)

plt.figure(figsize=(9, 7))
scatter = plt.scatter(X_tsne_2d[:, 0], X_tsne_2d[:, 1], c=y_digits, cmap="tab10", s=18, alpha=0.8)
plt.xlabel("t-SNE 1")
plt.ylabel("t-SNE 2")
plt.title("Digits dataset projected to 2D with t-SNE")
plt.legend(*scatter.legend_elements(), title="Digit")
plt.show()
```

t-SNE gives a 2D embedding that is usually much better at separating the digit classes visually because it preserves local neighborhood structure rather than maximizing global variance. In the t-SNE plot, the classes tend to form tighter and more distinct groups than in PCA.

### 3.4 ###
TSNE clusters the data better than PCA because the dataset has non linear structure so PCA can't fully capture this wth only 2 linear components. PCA is faster and simpler however, with more interpretable explained variance whle TSNE is good for visualization but requires more computation. That is why in scikit learn, TSNE usualy has fit_transform and not a separate transform, as it does not provide a true transform for new points. The embedding comes from the full dataset so adding new points isn't added in the same way that PCA does.
